# 🌊 Spark Structured Streaming Real-Time Processing

**Stream Processing over Distributed Storage**  
This notebook demonstrates how to initialize an Apache Spark Structured Streaming micro-batch query that continuously monitors an HDFS streaming ingestion folder (`hdfs://localhost:9000/stream-input`), aggregates windowed metrics, and outputs to the console.

```python
# Step 1: Initialize Spark Session with Streaming Enabled
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, DoubleType, TimestampType
from pyspark.sql.functions import col, window, current_timestamp

spark = SparkSession.builder \
    .appName("PySpark-StructuredStreaming-Simulation") \
    .master("local[*]") \
    .config("spark.hadoop.fs.defaultFS", "hdfs://localhost:9000") \
    .config("spark.sql.streaming.checkpointLocation", "/tmp/streaming-checkpoints") \
    .getOrCreate()

spark.sparkContext.setLogLevel("WARN")
print("⚡ Spark Structured Streaming Context Ready!")
```

```python
# Step 2: Define Schema for Incoming Stream Events
event_schema = StructType([
    StructField("device_id", StringType(), True),
    StructField("metric_name", StringType(), True),
    StructField("value", DoubleType(), True),
    StructField("timestamp", TimestampType(), True)
])

print("Stream Schema Defined:")
for field in event_schema.fields:
    print(f"  - {field.name} ({field.dataType.simpleString()})")
```

```python
# Step 3: Architecture of Ingestion Stream
print("=== Streaming Architecture Pattern ===")
print("1. Ingestion: IoT devices append micro-batch JSON/CSV events to HDFS /stream-input/")
print("2. Reader: spark.readStream.schema(event_schema).json('hdfs://localhost:9000/stream-input/')")
print("3. Transformation: Windowed 1-minute tumbling aggregations")
print("4. Sink: Output to HDFS Parquet Lakehouse or real-time console dashboard")
spark.stop()
```
